```python
df_titanic = sns.load_dataset('titanic')
df_titanic.sample()
```

```python
# sns.countplot
# sns.barplot
# sns.boxplot
# sns.scatterplot
# sns.catplot
# sns.pairplot
# sns.heatmap
# sns.relplot
# sns.lmplot
# sns.lineplot
# sns.kdeplot
# sns.violinplot
#sns.jointplot
```

```python
# sns.countplot
import matplotlib.pyplot as plt
sns.countplot(data = df_titanic,x= 'sex',hue = 'alive',palette=sns.color_palette('cool',2)) #hue: further divide:based on categories
plt.title('Titanic survival Analysis Based on gender!')
plt.show()
```

```python
# sns.barplot

sns.barplot(data = df_titanic,x = 'class',y = 'fare', hue = 'alive',estimator='mean') # min,max,sum
plt.show()
# these lines are called Confidence Interval(CI)
```

```python
# sns.barplot

sns.barplot(data = df_titanic,x = 'class',y = 'fare', hue = 'alive',errorbar= ('ci',0)) # min,max,sum
plt.grid(color = 'r',alpha = 0.3)
plt.show()
# these lines are called Confidence Interval(CI)
```

```python
# sns.boxplot:To detect Outliers and to check data distribtion
# To detect Outliers
# Range: Min to Max(19-22):35,28,45(Outliers: filter)

# Q2: Quantile : 50%
# Q1: 25%
# Q3: 75%

# IQR = Q3-Q1
# UW(Upper cap) = Q3+1.5*IQR
# LW(Lower cap) = Q1-1.5*IQR
```

```python
#sns.boxplot
sns.boxplot(data = df_titanic,x= 'class',y = 'fare', hue= 'sex')
plt.show()
```

```python
#sns.boxplot

sns.boxplot(data = df_titanic,y= 'class',x = 'fare', hue= 'sex', orient='h')
plt.show()
```

```python
#sns.violinplot
sns.violinplot(data = df_titanic,x= 'class',y = 'fare', hue= 'sex', orient='v')
plt.show()
```

```python
df_pen = sns.load_dataset('penguins')
df_pen.sample()
```

```python
df_tips = sns.load_dataset('tips')
df_tips.sample()
```

```python
# sns.catplot

sns.catplot(data = df_tips,x= 'time',y= 'total_bill',kind = 'bar',hue ='sex',row = 'smoker',col='day') #kind= strip
plt.show()
```

```python
sns.catplot(data = df_tips,x= 'time',y= 'total_bill',kind = 'violin',hue ='sex',row = 'smoker',col='day') #kind= strip
plt.show()
```

```python
# sns.lmplot
sns.lmplot(data = df_tips,x='total_bill',y='tip',row = 'smoker',col='day')
plt.show()
```

```python
# sns.relplot
sns.relplot(data = df_tips,x='total_bill',y='tip',kind='scatter',row = 'smoker',col='day',hue='sex')
plt.show()
```

```python
#sns.pairplot(check relationship pairwise)(num vs num)

df_iris = sns.load_dataset('iris')
df_iris.sample(5)
```

```python
df_iris.iloc[:,:-1]
```

```python
df_iris.iloc[:,:-1].corr()
```

```python
# >0: +ve : +ve corr
# =0: No corr
# <0: -ve : -ve corr
```

```python
sns.pairplot(data = df_iris)
plt.show()
```

```python
sns.pairplot(data = df_iris,hue ='species')
plt.show()
```

```python
sns.pairplot(data = df_iris,hue ='species',corner =True )
plt.show()
```

```python
#sns.kdeplot: kernel density estimation

sns.kdeplot(data = df_iris,x='sepal_length',y='petal_width')
plt.show()
```

```python
sns.kdeplot(data = df_iris,x='sepal_length',y='petal_width',hue='species')
plt.show()
```
