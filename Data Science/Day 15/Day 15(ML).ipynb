```python
data = {'Study_hrs': range(1,11),
       'Marks_Obtained':[25,33,46,50,55,67,78,85,95,100]}

df = pd.DataFrame(data)
df
```

```python
# ML model build?
# Scatter plot

plt.title('Study VS Marks')
plt.scatter(df['Study_hrs'],df['Marks_Obtained'],marker = 'x',color ='r')
plt.xlabel('Study Hrs')
plt.ylabel('Marks')
plt.show()
```

```python
# Import Module
from sklearn.linear_model import LinearRegression

# Object Model Creation
model = LinearRegression()

#x,y
x = df[['Study_hrs']]
y = df['Marks_Obtained']

# Training/Learning
model.fit(x,y)
```

```python
x = [[5.5]]

m = model.coef_
c = model.intercept_

print(m,c)
```

```python
# y = m*x+c
std_hrs = [[5.5]]

marks = m*std_hrs +c

print('Predicted Marks:',marks[0][0])
```

```python
import warnings
warnings.filterwarnings('ignore')

Study_hrs = input('Enter study hours:')
std_hr = [[float(Study_hrs)]]

ans = model.predict(std_hr)
marks = round(ans[0],2)

if marks > 100:
    marks = 100
    
elif marks < 0:
    marks = 0

else:
    marks = marks


print(f'Your Predicated marks is:{marks}')
```
